  ### **Task 1: Write Sales Records to a File**

   
    Create a list of sales amounts:   sales = [1200, 450, 980, 1500, 3000]   
    Open a file named sales_data.txt in write mode.   
    Write each sale on a new line in the file.   
    Close the file, then reopen it and print its contents.
    Extra (optional): Write the data in comma-separated format instead of separate lines. 

```python
# create a list of Sales amounts
sales = [1200, 450, 980, 1500, 3000]

# Open a file named sales_data.txt in write mode.
sales_data = open("sales_data.txt","w")

# Write each sale on a new line in the file.  
for amount in sales:
    sales_data.write(f"{amount}\n")

#Close the file, then reopen it and print its contents.
sales_data.close()
sales_read = open('sales_data.txt','r')
sales_content = sales_read.read()
print(f"Sales content: \n {sales_content}")
sales_data.close()
```

```python
# Extra (optional): Write the data in comma-separated format instead of separate lines. 

# create a list of Sales amounts
sales = [1200, 450, 980, 1500, 3000]

# comma-separated
comma_separated_sales = ",".join(map(str, sales))

# Open a file named sales_data.txt in write mode.
sales_data = open("sales_data.txt","w")

# Write the formatted string into the file
sales_data.write(comma_separated_sales)
sales_data.close()

sales_read = open("sales_data.txt","r")
sales_read.read()
```

## **Task 2: Read File in Different Ways**

    Using the same sales_data.txt:   
    Read the entire file using .read() and print it.   
    Read the first line using .readline().    
    Read all lines using .readlines() and convert them into a list of integers.
   

    Ensure proper formatting and cleanup of newline characters

```python
sales = [1200, 450, 980, 1500, 3000]

# Write each sale on its own line
with open("sales_data.txt", "w") as sale_write:
    for Salevalue in sales:
        sale_write.write(f"{Salevalue}\n")

# Read the entire file using .read()
with open("sales_data.txt", "r") as sale:
    content = sale.read()
    print("Full file content:\n", content)

# Read the first line using .readline()
with open("sales_data.txt", "r") as file:
    first_line = file.readline()
    print("First line:", first_line.strip())

# Read all lines using .readlines() and convert them into a list of integers
with open("sales_data.txt", "r") as sale_read:
    lines = sale_read.readlines()
    sales_list = [int(line.strip()) for line in lines]   # ✅ store in a list

# Print each integer
for value in sales_list:
    print(value)

print(f"Sales as integers: {sales_list}")

```

#  **Task 4: Generate Summary Report from File**

    Using only file reading operations:   
    Read all sales values from sales_data.txt.   
    Convert them into integers.   
    Calculate and print:   Total Sales   Highest Sale   Lowest Sale   Average Sale   

```python
with open("sales_data.txt","r") as readsales:
    lines = readsales.readlines()
    print("All sales Value")
    for line  in lines:
        print(int(line.strip()))
        

    # Convert each line into an integer
    sales_values = [int(line.strip()) for line in lines if line.strip()]

    # total Value of Sales
    print("-----------------------------------------")
    total_sales = sum(sales_values)
    print(f"total Sales : {total_sales}")
    print("-----------------------------------------")

    Highest_sales = max(sales_values)
    print(f"Highest Sales : {Highest_sales}")
    print("-----------------------------------------")

    Lowest_sales  = min(sales_values)
    print(f"lowest Sales : {Lowest_sales}")
    print("-----------------------------------------")

    Average_Sales = total_sales/ len(sales_values)
    print(f"Average Sales  : {Average_Sales}")
    print("-----------------------------------------")
```

# **Task 6: Read File Safely (Error Handling Inside File Handling Only)**


    You must not use exceptions beyond file-related safeguards here.
    Ask the user for a filename to open.
    If the file exists, read and print it.
    If it does not exist, print:File not found. Please check the filename.
    Use simple condition checks with os.path.exists() (allowed).

```python
import os


# Step 1: Ask user for a filename
filename = input("Enter the filename to open: ")

# Step 2: Check if file exists
if os.path.exists(filename):
    # Step 3: If exists, read and print content
    with open(filename, "r") as file:
        print("\nFile Content:\n")
        for line in file:
            print(line.strip())
else:
    # Step 4: If not exists, show message
    print("File not found. Please check the filename.")
```

# **Task 7: Mini Project – Export Discounted Prices**
    Objective:  
    Build a program that calculates discounted prices for products and exports them into a file.
    1. Create a dictionary with product names and their original prices.
    Mouse: 500
    Keyboard: 800
    Monitor: 7000
    Pendrive: 400
    Camera: 5000
    2. Ask the user to input a discount percentage.

    3. Calculate discounted prices using the formula: 
    discounted_price = original_price - (original_price * discount / 100)

    4. Write results into a file (discount_report.txt) in the format:Product | Original Price | Discounted Price

    5. Add a summary at the bottom of the file:
    Total Items: X
    Average Discounted Price: Y
 
    6. Read the file and print its contents to the terminal.



```python

# Step 1: Create dictionary
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

# Step 2: Ask user for discount percentage
discount = float(input("Enter discount percentage: "))

# Step 3: Write discounted prices into discount_report.txt
with open("discount_report.txt", "w") as report:
    report.write("Product | Original Price | Discounted Price\n")
    report.write("--------------------------------------------\n")

    discounted_values = []
    for product, price in prices.items():
        discounted_price = price - (price * discount / 100)
        discounted_values.append(discounted_price)
        report.write(f"{product} | {price} | {discounted_price:.2f}\n")

    # Optional summary
    total_items = len(prices)
    avg_discounted_price = sum(discounted_values) / total_items
    report.write("--------------------------------------------\n")
    report.write(f"Total Items: {total_items}\n")
    report.write(f"Average Discounted Price: {avg_discounted_price:.2f}\n")

# Step 4: Read and print the file
print("\nDiscount Report:\n")
with open("discount_report.txt", "r") as report:
    print(report.read())
```
